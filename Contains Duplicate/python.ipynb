```python
# calculate digit sum of a number
input = 19124
add = 0
while input > 0:
    digit = input % 10
    add = add + digit
    input = input // 10

print(f"The sum of digits is: {add}")
```

```python
# number reverse
input  = 1234

reverse = 0

while input > 0:
    digit = input % 10
    reverse = reverse * 10 + digit
    input = input // 10

print(f"The reverse of the number is: {reverse}")
```

```python
import json

OUTPUT_FILE = "answer.json"

questions = ["what is ai","what is python", "what is Json"]

def fake_ai_answer(question):
    return f"this is a sample answer to : {question}"

results = []

for question in questions:
    try:
        answer = fake_ai_answer(question)
        results.append({"question":question,"answer":answer})
        print(f"done {question} ?")
    except Exception as error:
        print(f"failed on {questions}:{error}")

with open(OUTPUT_FILE, "w") as f:
    json.dump(results,f, indent = 2)

print(f"\n all done! Saved {len(results)} answers to {OUTPUT_FILE}")

```
